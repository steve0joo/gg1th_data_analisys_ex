## 그래프에 한글표현

- WSL Ubuntu에서 한글 폰트 깨짐현상 해결

```python
# matplotlib 한글 폰트 설정 방법
# sudo apt update
# sudo apt install fonts-nanum
# fc-list | grep Nanum

# 그래도 폰트가 계속 깨진다면 아래 명령어 실행하기 
# rm -rf ~/.cache/matplotlib
```

```python
# plt.rcParams["font.family"] = 'Malgun Gothic'  # 윈도우
plt.rcParams['font.family'] = 'NanumGothic'      # 우분투
plt.rcParams['axes.unicode_minus'] = False
```

```python
plt.figure(figsize=(7, 5))
total = data.sum()
wedges, texts, autotexts = plt.pie(
  data, startangle=90, 
  wedgeprops={'width': 0.7},
  autopct=lambda pct: f'{round(pct/100*total):,}\n({pct:.1f}%)',
  pctdistance=0.82)
plt.setp(autotexts, size=9, weight='bold', color='white')
plt.title('게임세대별 갯수')

# loc: 범례의 기준점 / bbox_to_anchor: 축(axes) 좌표계 상의 부착 위치
# (1.0, 0.5) = 축의 오른쪽 끝, 세로 중앙  -> 범례가 차트 바깥에 놓인다
plt.legend(wedges, data.index, loc='center left', bbox_to_anchor=(1.0, 0.5),
          fontsize=10, title='세대별 갯수')

plt.tight_layout()
plt.savefig('x.png', dpi=100, bbox_inches='tight')
plt.show()
```

## Scatterplots

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
# 두 컬럼간의 관계를 알고 싶을때.
# 관계 => 상관관계분석   : 피어슨 상관계수 =>  -1 ~ 1 사이의 숫자
# 0 근처는 두 컬럼간에 아무 관계 없다.
# 1 로 갈수록 비례관계다 
# -1 로 갈수록 반비례 관계다. 
```

```python
df = pd.read_csv('./datas/fuel_econ.csv')
```

```python
# 상관분석 
df.corr(numeric_only= True)
```

```python
# 배기량과 연비의 관계 => 강한 반비례 관계 
df[  ['displ' , 'comb'] ].corr()
```

```python
df
```

```python
plt.scatter(data= df, x= 'displ', y= 'comb' )
plt.title('배기량과 연비의 관계')
plt.xlabel('displ')
plt.ylabel('comb')
plt.show()
```

```python
# 이 차트를 보고, 데이터는 3929개의 데이터 점을 보고서, fitting 되는 선(regression 회귀 )을 상상하면, 오른쪽 아래로 내려가는 
# 선을 그릴수 있다. 따라서 반비례 관계다 라고 알수 있다. 
```

```python
df
```

```python
# fitting 되는 선을 찾는것 => regression
```

```python
sb.regplot(data= df, x='displ', y='comb')
plt.show()
```

```python
# 상관계수도 같이 보여준다. 
df[['displ','comb']].corr()
```

```python
sb.pairplot(data= df,  vars= ['displ', 'comb', 'co2' ] )
plt.show()
```

## 문제 2. 엔진 크기와 이산화탄소 배출량의 관계를 분석하세요. 
- (히트맵으로)   displ 이 엔진사이즈이고, co2 가 이산화탄소 배출량입니다.  트랜드를 분석하세요

```python
df.columns
```

```python
#  엔진 크기와 이산화탄소 배출량의 관계 분석하기
# (히트맵으로)   displ 이 엔진사이즈이고, co2 가 이산화탄소 배출량입니다.  트랜드 분석하기
sb.heatmap(
    data= df[['displ','co2']].corr(), 
    annot= True , 
    fmt= '.2f', 
    cmap= 'coolwarm')
plt.title('엔진 크기와 이산화탄소 배출량의 관계')
```
